<h1>Creating a Pandas Series</h1>
A Pandas Series is like a single column of data in a spreadsheet. It is a one-dimensional array that can hold many types of data such as numbers, words or even other Python objects. Each value in a Series is associated with an index, which makes data retrieval and manipulation easy.

```python
import pandas as pd
import numpy as np
```

<h2>Creating a Series from a List</h2>

```python
# Pandas automatically assigns an index to each element starting from 0
data_list = ['a', 's', 'h', 'e', 'k']

ser = pd.Series(data_list)
ser
```

<h2>Creating a Series Using NumPy Functions</h2>

```python
# numpy.linspace(), numpy.random.randn()

ser = pd.Series(np.linspace(2, 10, 4))
ser
```

<h2>Creating a Series Using range()</h2>

```python
ser = pd.Series(range(5, 16, 5))
ser
```

<h2>Creating a Series Using List Comprehension</h2>

```python
ser = pd.Series(range(1, 21, 5), index=[x for x in 'abcd'])
ser
```

<h1>Accessing elements of a Pandas Series</h1>

Pandas Series is a one-dimensional labeled array capable of holding data of any type (integer, string, float, python objects, etc.).

Elements of a series can be accessed in two ways:

* Accessing Element from Series with Position
* Accessing Element Using Label (index)

<h2>Accessing Element from Series with Position</h2>

In order to access the series element refers to the index number. Use the index operator [ ] to access an element in a series. The index must be an integer.


```python
data = np.array(['g', 'e', 'e', 'k', 's', 'f',
                 'o', 'r', 'g', 'e', 'e', 'k', 's'])

ser = pd.Series(data)
ser[0] # Accessing Element from Series with Position
```

```python
ser[:5] # Accessing First 5 Elements of Series
```

```python
ser[-10 :] # Accessing Last 10 Elements of Series
```

```python
df = pd.read_csv('nba.csv')
```

```python
ser = pd.Series(df['Name'])
ser.head(6)
```

```python
ser[:5] # Accessing First 5 Elements of Series in nba.csv File
```

<h2>Access an Element in Pandas Using Label</h2>

```python
ser[2] # Accessing a Single Element Using index Label
```

```python
ser[[1, 2, 3, 4, 9]] # Accessing a Multiple Element Using index Label
```

```python
ser = pd.Series(np.arange(3, 9), index=['a', 'b', 'c', 'd', 'e', 'f'])
ser[['a', 'd']] # Access Multiple Elements by Providing Label of Index
```

```python
ser = pd.Series(df['Name'])
ser.head(4) # Accessing a Multiple Element Using Index Label in nba.csv File
```

<h1>Binary operations on Pandas DataFrame and Series</h1>

The most common binary operations include:

* Arithmetic operations: Addition, subtraction, multiplication, division, etc.
* Comparison operations: Equal to, not equal to, greater than, less than, etc.
* Logical operations: And, or, etc.

<h2>Binary Operations on Pandas Series</h2>

<h3>1. Arithmetic Operations on Series</h3>

Arithmetic operations between two Series is applied element-wise. The index labels must align for the operation to work. If the indexes don’t match, Pandas will fill in missing values with NaN.

```python
s1 = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
s2 = pd.Series([1, 2, 3], index=['a', 'b', 'c'])

result = s1 + s2
result
```

<h3>2. Comparison Operations on Series</h3>

Comparison operations return a Series of boolean values, indicating whether the comparison is True or False for each corresponding element.

```python
s1 = pd.Series([10, 20, 30])
s2 = pd.Series([10, 25, 30])

result = s1 == s2
result
```

<h2>Binary Operations on Pandas DataFrame</h2>

<h3>1. Arithmetic Operations on DataFrames</h3>

Similar to Series, DataFrame arithmetic operations apply element-wise between two DataFrames.

**Note:** The DataFrames must have the same shape or matching indexes and columns.

```python
df1 = pd.DataFrame({'A': [10, 20, 30], 'B': [40, 50, 60]})
df2 = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})

res = df1 - df2
res
```

<h3>2. Comparison Operations on DataFrames</h3>

Comparison operations on DataFrames return a DataFrame of boolean values.

```python
res = df1 > df2
res
```

<h2>Logical Operations on DataFrame and Series</h2>

Pandas also supports logical operations (AND, OR, etc.) on DataFrames and Series. These are commonly used for filtering and applying conditions.

```python
s1 = pd.Series([True, False, True])
s2 = pd.Series([False, False, True])

res = s1 & s2
res
```

<h2>Handling Missing Data in Binary Operations</h2>

* Arithmetic operations involving NaN will generally return NaN (e.g., NaN + 1 = NaN).
* Logical operations involving NaN might return False or True, depending on the operation.

```python
df1 = pd.DataFrame({'A': [1, 2, None], 'B': [4, None, 6]})
df2 = pd.DataFrame({'A': [1, None, 3], 'B': [None, 5, 6]})

res = df1 + df2
res
```

<h1>Pandas Series Index Attribute</h1>

```python
data = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])

print("Original index: ", data.index)

# Modifying the index
data.index = ['w', 'x', 'y', 'z']
print("Modified Series:\n", data)
```

<h3>Example 1. Assigning Duplicate Index Labels</h3>

```python
series = pd.Series(['New York', 'Chicago', 'Toronto', 'Lisbon'])

series.index = ['City 1', 'City 1', 'City 3', 'City 3']
series # Pandas allows assigning duplicate index labels
```

<h3>Example 2. Retrieving Index Labels</h3>

```python
import pandas as pd

Date = ['1/1/2018', '2/1/2018', '3/1/2018', '4/1/2018']
idx_name = ['Day 1', 'Day 2', 'Day 3', 'Day 4']

ser = pd.Series(data=Date, index=idx_name)
ser.index
```

<h3>Example 3. Resetting Index to Default</h3>

```python
sr = pd.Series(data=Date, index=idx_name)

sr.reset_index(drop=True, inplace=True)
sr
```
